Red neuronal profunda entrenada para clasificar automáticamente **10 categorías de prendas de vestir** usando el dataset Fashion MNIST.

**Arquitectura:** `Flatten (28×28) → Dense 128 (ReLU) → Dropout 0.2 → Dense 64 (ReLU) → Dense 10 (Softmax)`  
**Precisión final:** 87.72%

## ⚙️ Paso 1 — Importaciones y verificación de TensorFlow

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
import matplotlib.pyplot as plt

print("✅ TensorFlow versión:", tf.__version__)
```

## 📦 Paso 2 — Carga del dataset Fashion MNIST

Fashion MNIST es un dataset de 70,000 imágenes en escala de grises (28×28 píxeles) distribuidas en 10 categorías de ropa. Viene incluido en TensorFlow — no hay que descargar nada externamente.

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.fashion_mnist.load_data()

# Nombres de las categorías
categorias = [
    "Camiseta", "Pantalón", "Suéter", "Vestido", "Abrigo",
    "Sandalia", "Camisa", "Tenis", "Bolso", "Botín"
]

print(f"✅ Datos de entrenamiento: {X_train.shape}")
print(f"✅ Datos de prueba: {X_test.shape}")
```

## 🖼️ Paso 3 — Visualización de ejemplos del dataset

```python
plt.figure(figsize=(10, 5))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_train[i], cmap='gray')
    plt.title(categorias[y_train[i]])
    plt.axis('off')
plt.suptitle('Ejemplos del Dataset')
plt.tight_layout()
plt.show()
```

## 🔧 Paso 4 — Normalización de los datos

Los valores de píxeles van de 0 a 255. Se normalizan al rango [0, 1] dividiendo entre 255, lo que acelera el entrenamiento y mejora la convergencia.

```python
X_train = X_train / 255.0
X_test = X_test / 255.0

print(f"✅ Datos normalizados: rango [{X_train.min():.1f}, {X_train.max():.1f}]")
```

## 🧠 Paso 5 — Construcción de la red neuronal

Se construye una red neuronal densa con las siguientes capas:
- **Flatten**: convierte la imagen 28×28 en un vector de 784 valores
- **Dense 128 (ReLU)**: primera capa oculta con 128 neuronas
- **Dropout 0.2**: regularización para evitar overfitting
- **Dense 64 (ReLU)**: segunda capa oculta con 64 neuronas
- **Dense 10 (Softmax)**: capa de salida con 1 neurona por categoría

```python
modelo = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])

modelo.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

modelo.summary()
```

## 🏋️ Paso 6 — Entrenamiento del modelo

El modelo se entrena durante 10 épocas con un 10% de los datos reservados para validación. El entrenamiento tarda aproximadamente **2-3 minutos** en Google Colab.

```python
historia = modelo.fit(
    X_train, y_train,
    epochs=10,
    validation_split=0.1,
    verbose=1
)
```

## 📊 Paso 7 — Evaluación y curvas de aprendizaje

```python
loss, accuracy = modelo.evaluate(X_test, y_test, verbose=0)
print(f"✅ Precisión en datos de prueba: {accuracy:.2%}")

# Curvas de aprendizaje
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(historia.history['accuracy'], label='Entrenamiento')
ax1.plot(historia.history['val_accuracy'], label='Validación')
ax1.set_title('Precisión por época')
ax1.set_xlabel('Época')
ax1.set_ylabel('Precisión')
ax1.legend()

ax2.plot(historia.history['loss'], label='Entrenamiento')
ax2.plot(historia.history['val_loss'], label='Validación')
ax2.set_title('Pérdida por época')
ax2.set_xlabel('Época')
ax2.set_ylabel('Pérdida')
ax2.legend()

plt.tight_layout()
plt.show()
```

## 🔍 Paso 8 — Predicciones de ejemplo

```python
predicciones = modelo.predict(X_test[:10], verbose=0)

plt.figure(figsize=(15, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(X_test[i], cmap='gray')
    pred = categorias[np.argmax(predicciones[i])]
    real = categorias[y_test[i]]
    color = 'green' if pred == real else 'red'
    plt.title(f"Pred: {pred}\nReal: {real}", color=color, fontsize=8)
    plt.axis('off')
plt.suptitle('Predicciones (verde=correcto, rojo=incorrecto)')
plt.tight_layout()
plt.show()
```

## 📋 Paso 9 — Resumen del proyecto

```python
print("=" * 50)
print("RESUMEN DEL PROYECTO")
print("=" * 50)
print(f"Dataset: Fashion MNIST")
print(f"Total imágenes entrenamiento: {len(X_train)}")
print(f"Total imágenes prueba: {len(X_test)}")
print(f"Categorías clasificadas: {len(categorias)}")
print(f"Precisión final: {accuracy:.2%}")
print()
print("El modelo de Red Neuronal clasifica correctamente")
print(f"aproximadamente {accuracy:.0%} de las imágenes de ropa.")
```
